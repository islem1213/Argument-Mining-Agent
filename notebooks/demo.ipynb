# Argument-mining walkthrough
Mine the bundled essay, inspect components and relations, then look at what formal argumentation semantics say about it.

```python
from pathlib import Path

from argmine import ArgumentMiner, PipelineConfig
from argmine.export import render_html, summary, to_mermaid

text = Path('../examples/school_start_times.txt').read_text()
graph = ArgumentMiner(PipelineConfig(fallacies='heuristic')).mine(text)
print(summary(graph))
```

Grounded semantics and DF-QuAD strength per component, plus diagnostics:

```python
graph.analysis.model_dump()
```

```python
from IPython.display import HTML

HTML(render_html(graph, text))
```

```python
print(to_mermaid(graph))
```
